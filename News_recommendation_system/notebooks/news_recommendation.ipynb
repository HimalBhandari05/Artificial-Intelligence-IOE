## Loading Datasets and Librarires 

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.metrics import accuracy_score , recall_score , f1_score , precision_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfTransformer,TfidfVectorizer
from sklearn.pipeline import make_pipeline
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# loading the actual data. 

news = pd.read_csv("../data/bbc_news_dataset.csv")
news
```

```python
# ready the data here. 

X = news["Text"]
y = news["Category"]
```

```python
# model pipeline and training it.

vectorizer = TfidfVectorizer()

tf_idf_matrix = vectorizer.fit_transform(X)
tf_idf_matrix
```

```python
# user enters a query 

query_vector = vectorizer.transform(["apple ai"])
query_vector
```

```python
# now similarity is measured. and recommendation can be made. 

similarity = cosine_similarity(query_vector , tf_idf_matrix)
similarity
```

```python
# since it is an array we can sort it .

indices = similarity.argsort()[0][-3:][::-1]
indices
```
